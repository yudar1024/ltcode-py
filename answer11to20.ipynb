````python
from typing import List

"""
11. 除自身以外数组的乘积
12. Integer to Roman
13. Roman to Integer
14. Longest Common Prefix
15. 3Sum
16. 3Sum Closest
17. Letter Combinations of a Phone Number
18. 4Sum
19. Remove Nth Node From End of List
20. Valid Parentheses
"""
def productExceptSelf(nums: List[int]) -> List[int]:
    """
    11. 除自身以外数组的乘积，不要使用除法，且在 O(n) 时间复杂度内完成此题。
    示例 1:
    输入: nums = [1,2,3,4]
    输出: [24,12,8,6]
    """
    n = len(nums)
    res = [1] * n
    print(f"res={res}")
    # 计算左侧乘积
    for i in range(1, n):
        print(f"计算左侧乘积 i={i}, res[i-1]={res[i-1]}, nums[i-1]={nums[i-1]}")
        res[i] = res[i-1] * nums[i-1]
    right = nums[-1]
    print(f"right={right}")
    # 计算右侧乘积
    for i in range(n-2, -1, -1):
        print(f"计算右侧乘积 i={i}, res[i]={res[i]}, nums[i]={nums[i]}")
        res[i] *= right
        right *= nums[i]
    return res
````

````python
productExceptSelf([1,2,3,4])
````

理解 左部分乘积与右部分乘积
```
原数组：       [1       2       3       4]
左部分的乘积：   1       1      1*2    1*2*3
右部分的乘积： 2*3*4    3*4      4      1
结果：        1*2*3*4  1*3*4   1*2*4  1*2*3*1
```

### 134 加油站

在一条环路上有 n 个加油站，其中第 i 个加油站有汽油 gas[i] 升。
你有一辆油箱容量无限的的汽车，从第 i 个加油站开往第 i+1 个加油站需要消耗汽油 cost[i] 升。你从其中的一个加油站出发，开始时油箱为空。
给定两个整数数组 gas 和 cost ，如果你可以按顺序绕环路行驶一周，则返回出发时加油站的编号，否则返回 -1 。如果存在解，则 保证 它是 唯一 的。

````python
def canCompleteCircuit(gas: List[int], cost: List[int]) -> int:
    """
    134. 加油站
    """
    n = len(gas)
    total_tank, curr_tank = 0, 0
    starting_station = 0
    for i in range(n):
        # 计算从当前加油站到下一个加油站的净油量,
        # 如果净油量为负,则说明从当前加油站出发无法到达下一个加油站
        # 因此,需要从下一个加油站出发
        total_tank += gas[i] - cost[i]
        # If one couldn't get to the next station, start from the next station.
        curr_tank += gas[i] - cost[i]
        if curr_tank < 0:
            starting_station = i + 1
            curr_tank = 0
            # 如果总油量小于总消耗,则说明无法完成环路行驶
            # 因此,返回 -1
    return starting_station if total_tank >= 0 else -1
````
