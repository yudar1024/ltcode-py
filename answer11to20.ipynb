````python
from typing import List

"""
11. 除自身以外数组的乘积
12. Integer to Roman
13. Roman to Integer
14. Longest Common Prefix
15. 3Sum
16. 3Sum Closest
17. Letter Combinations of a Phone Number
18. 4Sum
19. Remove Nth Node From End of List
20. Valid Parentheses
"""
def productExceptSelf(nums: List[int]) -> List[int]:
    """
    11. 除自身以外数组的乘积，不要使用除法，且在 O(n) 时间复杂度内完成此题。
    示例 1:
    输入: nums = [1,2,3,4]
    输出: [24,12,8,6]
    """
    n = len(nums)
    res = [1] * n
    print(f"res={res}")
    # 计算左侧乘积
    for i in range(1, n):
        print(f"计算左侧乘积 i={i}, res[i-1]={res[i-1]}, nums[i-1]={nums[i-1]}")
        res[i] = res[i-1] * nums[i-1]
    right = nums[-1]
    print(f"right={right}")
    # 计算右侧乘积
    for i in range(n-2, -1, -1):
        print(f"计算右侧乘积 i={i}, res[i]={res[i]}, nums[i]={nums[i]}")
        res[i] *= right
        right *= nums[i]
    return res
````

理解 左部分乘积与右部分乘积
```
原数组：       [1       2       3       4]
左部分的乘积：   1       1      1*2    1*2*3
右部分的乘积： 2*3*4    3*4      4      1
结果：        1*2*3*4  1*3*4   1*2*4  1*2*3*1
```

### 134 加油站

在一条环路上有 n 个加油站，其中第 i 个加油站有汽油 gas[i] 升。
你有一辆油箱容量无限的的汽车，从第 i 个加油站开往第 i+1 个加油站需要消耗汽油 cost[i] 升。你从其中的一个加油站出发，开始时油箱为空。
给定两个整数数组 gas 和 cost ，如果你可以按顺序绕环路行驶一周，则返回出发时加油站的编号，否则返回 -1 。如果存在解，则 保证 它是 唯一 的。

````python
def canCompleteCircuit(gas: List[int], cost: List[int]) -> int:
    """
    134. 加油站
    """
    n = len(gas)
    total_tank, curr_tank = 0, 0
    starting_station = 0
    for i in range(n):
        # 计算从当前加油站到下一个加油站的净油量,
        # 如果净油量为负,则说明从当前加油站出发无法到达下一个加油站
        # 因此,需要从下一个加油站出发
        total_tank += gas[i] - cost[i]
        # If one couldn't get to the next station, start from the next station.
        curr_tank += gas[i] - cost[i]
        if curr_tank < 0:
            starting_station = i + 1
            curr_tank = 0
            # 如果总油量小于总消耗,则说明无法完成环路行驶
            # 因此,返回 -1
    return starting_station if total_tank >= 0 else -1
````

### 135 分发糖果
n 个孩子站成一排。给你一个整数数组 ratings 表示每个孩子的评分。

你需要按照以下要求，给这些孩子分发糖果：

每个孩子至少分配到 1 个糖果。
相邻两个孩子中，评分更高的那个会获得更多的糖果。
请你给每个孩子分发糖果，计算并返回需要准备的 最少糖果数目 。



````python
def candy(ratings: List[int]) -> int:
    """
    135. 分发糖果
    """
    n = len(ratings)
    # 每个孩子至少分配到 1 个糖果
    candies = [1] * n
    # 从左到右遍历,如果当前孩子的评分大于前一个孩子的评分,
    # 则当前孩子的糖果数为前一个孩子的糖果数加 1
    for i in range(1, n):
        if ratings[i] > ratings[i - 1]:
            candies[i] = candies[i - 1] + 1
    # 从右到左遍历,如果当前孩子的评分大于后一个孩子的评分,
    # 则当前孩子的糖果数为后一个孩子的糖果数加 1    
    for i in range(n - 2, -1, -1):
        if ratings[i] > ratings[i + 1]:
            candies[i] = max(candies[i], candies[i + 1] + 1)
    return sum(candies)
````

### 136 只出现一次的数字
给你一个 非空 整数数组 nums ，除了某个元素只出现一次以外，其余每个元素均出现两次。找出那个只出现了一次的元素。

你必须设计并实现线性时间复杂度的算法来解决此问题，且该算法只使用常量额外空间。

````python
from typing import List
def singleNumber(nums: List[int]) -> int:
    res = 0
    # 异或操作，相同为0，不同为1
    # 0 ^ 任何数 = 任何数
    # 任何数 ^ 任何数 = 0
    for num in nums:
        # 先与0异或，再与其他数异或
        print(f"before: res: {res}, num: {num}, res^num: {res^num}")
        res ^= num
        print(f"after: res: {res}, num: {num}, res^num: {res^num}")
    return res

singleNumber([4,1,2,1,2])
````

````python
#!/bin/python3
# 给定一个包含字母、数字和符号的字符串，仅考虑字母字符时 (不区分大小写), 确定它是否正向和反向读取相同的内容。
#
# Complete the 'isAlphabeticPalindrome' function below.
#
# The function is expected to return a BOOLEAN.
# The function accepts STRING code as parameter.
#

def isAlphabeticPalindrome(code:str) -> bool:
    # Write your code here
    if len(code)==1:
        return True
    code = code.lower()
    left = 0
    right = len(code) - 1
    while left < right:
        if not code[left].isalpha():
            left += 1
        elif not code[right].isalpha():
            right -= 1
        elif code[left] != code[right]:
            return False
        else:
            left += 1
            right -= 1
    return True

isAlphabeticPalindrome("abc123cba")
````

````python
#!/bin/python3
#给定一个区间数组 [startTime，endTime], 合并所有重叠的区间并返回一个有序的非重叠区间数组。
# Example
#
# Input
#
# intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]
# Output
#
# [[1, 6], [8, 10], [15, 18]]
# Complete the 'mergeHighDefinitionIntervals' function below.
#
# The function is expected to return a 2D_INTEGER_ARRAY.
# The function accepts 2D_INTEGER_ARRAY intervals as parameter.
#

def mergeHighDefinitionIntervals(intervals):
    if not intervals:
        return []

    # Sort intervals by start time
    intervals.sort(key=lambda x: x[0])

    # Initialize merged list with first interval
    merged = [intervals[0]]

    # Iterate through remaining intervals
    for i in range(1, len(intervals)):
        current = intervals[i]
        last_merged = merged[-1]

        # Check if current interval overlaps with last merged interval
        if current[0] <= last_merged[1]:
            # Merge by updating the end time to the maximum
            merged[-1][1] = max(last_merged[1], current[1])
        else:
            # No overlap, append current interval
            merged.append(current)

    return merged
````
